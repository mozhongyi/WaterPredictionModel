```python
import pandas as pd
import os
from pathlib import Path

def split_water_level_excel(input_file, output_dir=None, time_column='时间'):
    """
    将包含多个水位点的Excel表格拆分为单独的Excel文件
    
    参数:
    input_file (str): 输入Excel文件路径
    output_dir (str, optional): 输出文件夹路径，默认为输入文件所在目录
    time_column (str, optional): 时间列的列名，默认为'时间'
    """
    # 读取Excel文件
    try:
        df = pd.read_excel(input_file)
    except Exception as e:
        print(f"读取文件失败: {e}")
        return
    
    # 检查时间列是否存在
    if time_column not in df.columns:
        print(f"错误: 找不到时间列 '{time_column}'")
        return
    
    # 确定输出目录
    if output_dir is None:
        output_dir = Path(input_file).parent
    else:
        os.makedirs(output_dir, exist_ok=True)
    
    # 获取所有水位列（除时间列外的所有列）
    water_level_columns = [col for col in df.columns if col != time_column]
    
    if not water_level_columns:
        print("错误: 没有找到水位列")
        return
    
    # 为每个水位列创建单独的Excel文件
    for column in water_level_columns:
        # 创建仅包含时间列和当前水位列的DataFrame
        new_df = df[[time_column, column]].copy()
        
        # 构建输出文件名
        output_file = os.path.join(output_dir, f"{column}.xlsx")
        
        try:
            # 保存到新的Excel文件
            new_df.to_excel(output_file, index=False)
            print(f"已保存: {output_file}")
        except Exception as e:
            print(f"保存文件 {output_file} 失败: {e}")
    
    print(f"处理完成! 共生成 {len(water_level_columns)} 个文件")
```

```python
input_file = "./OriginData/2025.1.1-5.31.xlsx"  # 替换为您的输入文件路径
output_dir = "./OutputData/"      # 替换为您的输出目录，默认为None表示使用输入文件所在目录

split_water_level_excel(input_file, output_dir)   
```
